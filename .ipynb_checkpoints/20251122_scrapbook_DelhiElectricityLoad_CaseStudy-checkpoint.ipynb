```python
import pandas as pd
import numpy as np
import os
from tqdm import tqdm
```

```python
### Making the archive_data.csv file by concating multiple csv files (Don't Include in the final code)

# Function to load a daily file, extract Delhi load, convert filename date, 
# create 15-min timestamps, and return a clean 96-row dataframe
def process_files(file_name):
    
    # Build full file path
    data_file_path = f'archive_data/{file_name}'
    
    # Load CSV and keep only Delhi load (first 96 rows)
    file_data = pd.read_csv(data_file_path)
    file_data = file_data[['Delhi']].head(96)
    
    # Extract and convert date from filename: DD-MM-YYYY → YYYY-MM-DD
    raw_date = file_name.replace('.csv', '').split('DATE')[1]
    file_date = pd.to_datetime(raw_date, format="%d-%m-%Y").strftime("%Y-%m-%d")
    
    # Create 96 timestamps at 15-minute intervals
    time_index = pd.date_range(start=file_date, freq='15min', periods=96)
    
    # Add timestamp column
    file_data['timestamp'] = time_index
    
    return file_data

# Get list of all files to be merged
file_names = os.listdir('archive_data')

# Create an empty master dataframe
archive_data = pd.DataFrame()

# Process each file and append results into the master dataframe
for i in tqdm(file_names, desc="Processing files"):
    try:
        file_df = process_files(file_name=i)
        archive_data = pd.concat([archive_data, file_df])
    except Exception:
        print("Skipped:", i)
archive_data['timestamp'] = archive_data['timestamp'].astype(str)
archive_data.to_csv('archive_data.csv', index = False)
print('__________Completed__________')
```
